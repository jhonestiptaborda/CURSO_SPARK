# 04 — Selección de columnas en PySpark

## 1. Objetivo

Aprender a seleccionar una o varias columnas de un DataFrame utilizando
`select()`.

En este módulo se estudiará:

- selección de una columna;
- selección de varias columnas;
- cambio del orden de las columnas;
- selección utilizando nombres de columnas;
- selección utilizando objetos `Column`;
- diferencia entre `select()` y `show()`.

La selección de columnas permite trabajar únicamente con los atributos
necesarios para una transformación o una regla de negocio.

## 2. Concepto oficial

El método `select()` proyecta un conjunto de columnas o expresiones y devuelve
un nuevo DataFrame.

Sintaxis general:

`df.select("columna1", "columna2")`

El DataFrame original no se modifica. El resultado de `select()` es otro
DataFrame que contiene únicamente las columnas seleccionadas.

```python
import os
import sys

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

print("Python utilizado:", sys.executable)
```

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("04_seleccion_columnas")
    .master("local[2]")
    .config("spark.python.worker.reuse", "false")
    .getOrCreate()
    )

print(spark.version)
    
```

```python
ruta_customers = "../data/raw/customers.csv"

df_customers = spark.read.csv(ruta_customers, header=True, inferSchema=True)
df_customers.show(3, truncate=False)
df_customers.printSchema()
```

## 3. Sintaxis de `select()`

El método `select()` permite escoger una o varias columnas de un DataFrame.

La sintaxis básica es:

`df.select("nombre_columna")`

Para seleccionar varias columnas:

`df.select("columna1", "columna2", "columna3")`

`select()` devuelve un nuevo DataFrame con las columnas seleccionadas.

El DataFrame original no se modifica.

```python
df_customer_ids = df_customers.select("customer_id")

df_customer_ids.show(5, truncate=False)
df_customer_ids.printSchema()
```

### Explicación del ejemplo

`df_customers`

Es el DataFrame original que contiene todas las columnas de clientes.

`select("customer_id")`

Selecciona únicamente la columna `customer_id`.

`df_customer_ids`

Es un nuevo DataFrame que contiene solamente la columna seleccionada.

`show(5, truncate=False)`

Muestra los primeros cinco registros sin recortar los valores.

`printSchema()`

Permite comprobar que el nuevo DataFrame contiene una sola columna.

### Resultado del ejemplo guiado

Se creó un nuevo DataFrame llamado `df_customer_ids` a partir de
`df_customers`.

Con `select("customer_id")` se seleccionó únicamente la columna
`customer_id`.

El DataFrame original no se modificó. El nuevo DataFrame quedó con una sola
columna de tipo `string`.

## 4. Selección de varias columnas

El método `select()` también permite seleccionar varias columnas.

Sintaxis:

`df.select("columna1", "columna2", "columna3")`

El orden en que se escriben las columnas dentro de `select()` será el orden
que tendrán en el nuevo DataFrame.

El DataFrame original no se modifica.

```python
df_customer_names = df_customers.select(
    "customer_id",
    "first_name",
    "last_name"
)

df_customer_names.show(5, truncate=False)
df_customer_names.printSchema()
```

### Resultado del ejemplo guiado 2

El nuevo DataFrame contiene tres columnas:

1. `customer_id`
2. `first_name`
3. `last_name`

Las columnas aparecen en el mismo orden en que fueron escritas dentro de
`select()`.

El DataFrame original `df_customers` no fue modificado.

## 5. Cambio del orden de las columnas

El orden de las columnas en el nuevo DataFrame depende del orden en que se
escriban dentro de `select()`.

Por ejemplo:

`df.select("city", "customer_id", "email")`

crea un nuevo DataFrame con las columnas en este orden:

1. `city`
2. `customer_id`
3. `email`

El contenido de las columnas no cambia; solamente cambia su posición dentro
del nuevo DataFrame.

```python
df_customer_contact_ordered = df_customers.select("city","customer_id","email")
df_customer_contact_ordered.show(5, truncate=False)
df_customer_contact_ordered.printSchema()
```

### Resultado del ejemplo guiado 3

La columna `city` quedó en la primera posición.

La columna `email` quedó en la última posición.

Respecto al ejercicio anterior, únicamente cambió el orden de las columnas.

Los datos no fueron modificados y el DataFrame original `df_customers`
permanece igual.

## 6. Selección utilizando objetos `Column`

Una columna también puede seleccionarse mediante esta sintaxis:

`df["nombre_columna"]`

Al utilizarla dentro de `select()`:

`df.select(df["nombre_columna"])`

`df["nombre_columna"]` representa un objeto `Column`.

Esta forma será útil más adelante para construir expresiones y
transformaciones sobre columnas.

```python
df_customer_ids_column = df_customers.select(
    df_customers["customer_id"]
)

df_customer_ids_column.show(5, truncate=False)
df_customer_ids_column.printSchema()
```

### Resultado del ejemplo guiado 4

1. `df_customers["customer_id"]` representa la columna llamada
   `customer_id` como un objeto `Column`.

2. El nuevo DataFrame tiene una sola columna.

3. No existe una diferencia visible en el resultado frente a
   `select("customer_id")`. La diferencia está en la forma de representar la
   columna dentro del código.

4. El DataFrame original no fue modificado.

## 7. Selección combinando texto y objetos `Column`

PySpark permite combinar ambas formas dentro del mismo `select()`:

`df.select("columna1", df["columna2"])`

Una columna puede pasarse como texto y otra como objeto `Column`.

Aunque ambas formas son válidas, conviene mantener un estilo consistente para
que el código sea más fácil de leer.

```python
df_customer_mixed = df_customers.select(
    "customer_id",
    df_customers["credit_score"],
    "created_at"
)

df_customer_mixed.show(5, truncate=False)
df_customer_mixed.printSchema()
```

### Resultado del ejemplo guiado 5

1. El nuevo DataFrame tiene 3 columnas.

2. `credit_score` se seleccionó como objeto `Column`.

3. `customer_id` y `created_at` se seleccionaron como texto.

4. El resultado visible no cambia por combinar ambas formas de selección.
